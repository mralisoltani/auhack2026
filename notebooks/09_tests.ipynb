```python
# Flows Check
import pandas as pd 

zones_f = ["AT", "BE","CH", "CZ", "DE", "DK1", "DK2", "FR", "NL", "NO2", "PL", "SE4"]
for z in zones_f:
    tmp_f = pd.read_csv(f"data/flows/{z}-physical-flows-in.csv")
    print(tmp_f.isnull().sum())
    print(tmp_f["time"].min())
    print(tmp_f["time"].max())
    print("Length: ", len(tmp_f))
    print(tmp_f.columns)


```

```python
# generations Check

zones_g = ["AT", "BE","CH", "CZ", "DE", "DK1", "FR", "NL", "PL"]
for z in zones_g:
    tmp_g = pd.read_csv(f"data/generation/{z}-generation.csv")
    print(tmp_g.head())
    print(tmp_g.isnull().sum())
    print(tmp_g["time"].min())
    print(tmp_g["time"].max())
    print(tmp_g.columns)
```

```python
# weather Check

import glob

zones_p = ["AT", "BE","CH", "CZ", "DE", "DK1", "FR", "NL", "PL"]
for z in zones_p:
    matches = glob.glob(f"data/weather/{z}-open-meteo-*.csv")
    if not matches:
        print(f"No weather file found for {z}")
        continue
    tmp_p = pd.read_csv(matches[0], skiprows=3)
    print(tmp_p.isnull().sum())
    print(tmp_p["time"].min())
    print(tmp_p["time"].max())
    print(tmp_p.columns)
```
